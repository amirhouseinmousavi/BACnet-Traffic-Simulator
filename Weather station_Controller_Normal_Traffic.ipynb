## Defined features for the controller
Analog Inputs:

- **Outdoor Air Humidity**
- **Outdoor Air Wet bulb Temp**
- **OutdoorTemp**



```python
# define device objects
def defining_objects(weather_station):
    # start from fresh
    ObjectFactory.clear_objects()
    
    #content of the object is defined here anward
    
    weather_station1 = analog_input(
    instance=80,
    name = "CP_WeatherStation1_OutdoorAirHumidity",
    properties = {"units" : "percent"},
    description = "CP_WeatherStation1_OutdoorAirHumidity",
    presentValue = 80.0,
    )
    analog_input(
        instance=25,
        name="CP_WeatherStation1_OutdoorAirWetbulbTemp",
        properties={"units" : "degreesCelsius"},
        description="CP_WeatherStation1_OutdoorAirWetbulbTemp",
        presentValue=25.0,
    )
    analog_input(
        instance=25,
        name="CP_WeatherStation1_OutdoorTemp",
        properties={"units" : "degreesCelsius"},
        description="CP_WeatherStation1_OutdoorTemp",
        presentValue=25.0,
    )
    
    return weather_station1.add_objects_to_application(weather_station)
```

Configuring the controller to connect to the network involves:
- **reading the network interface IP**
- **using it for controller simulation**
- **specifying the port for the back-net protocol**
- **defining an ID for the controller**

```python
adapters = ifaddr.get_adapters()
ip = None
for adapter in adapters:
    if adapter.nice_name == 'maineth':
        ip = adapter.ip_addresses[0] 
        break

weather_station = connect(ip=ip, port='47808', deviceId='3330')
defining_objects(weather_station)
```

```python
start_date = time.struct_time((2024, 2, 28, 16, 29, 10, 0, 0, 0))
start_timestamp = time.mktime(start_date)

for i in range(int(start_timestamp - time.time())):
  time.sleep(1)

with open('E://dataset_for_simulation/weather station1.csv') as f:
    reader = csv.reader(f)
    rows = list(reader)

row_num = 1 

while True:
    if row_num >= len(rows):
        break
        
    CP_WeatherStation1_OutdoorAirHumidity = float(rows[row_num][1])
    CP_WeatherStation1_OutdoorAirWetbulbTemp = float(rows[row_num][2])
    CP_WeatherStation1_OutdoorTemp = float(rows[row_num][3])

    

    weather_station["CP_WeatherStation1_OutdoorAirHumidity"].presentValue = CP_WeatherStation1_OutdoorAirHumidity 
    weather_station["CP_WeatherStation1_OutdoorAirWetbulbTemp"].presentValue = CP_WeatherStation1_OutdoorAirWetbulbTemp
    weather_station["CP_WeatherStation1_OutdoorTemp"].presentValue = CP_WeatherStation1_OutdoorTemp

    
    print("weather_station1 CP_WeatherStation1_OutdoorAirHumidity :", weather_station["CP_WeatherStation1_OutdoorAirHumidity"].presentValue) 
    print("weather_station1 CP_WeatherStation1_OutdoorAirWetbulbTemp :", weather_station["CP_WeatherStation1_OutdoorAirWetbulbTemp"].presentValue)   
    print("weather_station1 CP_WeatherStation1_OutdoorTemp :", weather_station["CP_WeatherStation1_OutdoorTemp"].presentValue) 

    row_num += 1
    sleep(2)

print("Reached end of data")
```
